```python
import pandas as pd
df = pd.read_csv('../data/application_tr.csv')
memory_mb = df.memory_usage(deep=True).sum() / 1024**2
print(f"Исходный размер датасета в памяти: {memory_mb:.2f} MB")
```

```python

for col in df.select_dtypes(include=['float64']).columns:
    df[col] = pd.to_numeric(df[col], downcast='float')

for col in df.select_dtypes(include=['int64']).columns:
    df[col] = pd.to_numeric(df[col], downcast='integer')

for col in df.select_dtypes(include=['object']).columns:
    if df[col].nunique() / len(df) < 0.5: 
        df[col] = df[col].astype('category')

mem_after = df.memory_usage(deep=True).sum() / 1024**2
print(f"Оптимизированный размер: {mem_after:.2f} MB")
df.to_parquet('../data/application_optimized.parquet', engine='pyarrow')
print("Успешно сохранено в Parquet!")
```
